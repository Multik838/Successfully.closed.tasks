## Задания № 1:
### Топ-5 фамилий менеджеров по доли успешно закрытых задач

```python
# От себя добавлю, что нам требуется вывести "топ 5 фамилий менеджеров по доли успешно закрытых задач"
```

```python
# Т.е. кол-во успешно выполненных задач у USER_ID к общему кол-ву задач USER_ID
```

```python
users_analysis = df[['USER_ID', 'TASK_ID', 'STATUS', 'LAST_NAME']]
```

```python
# Подсчитаем долю пользователей, прослушавших композиции стиля funk:
df_success_users = users_analysis.query('STATUS == "success"')[['USER_ID', 'TASK_ID', 'STATUS', 'LAST_NAME']]

# Переименуем столбец TASK_ID, чтобы не запустаться в данных
df_success_users.rename(columns={'TASK_ID': 'SUCCESS_TASK_ID'}, inplace=True)

df_success_users.reset_index()
```

```python
print(df_success_users)
```

```python
# Посчитаем кол-во успешных задач для каждого пользователя
df_success_users = df_success_users.groupby('USER_ID').agg({'SUCCESS_TASK_ID':'count'})
print(df_success_users)
```

```python
# Посчитаем общее кол-во задач для каждого user_id
df_total_users = users_analysis.groupby('USER_ID').agg({'TASK_ID':'count'})
# Переименуем столбец TASK_ID, чтобы не запустаться в данных
df_total_users.rename(columns={'TASK_ID': 'TOTAL_TASK_ID'}, inplace=True)

df_total_users.reset_index()

print(df_total_users)
```

```python
# Или с указанием типа объединения
joined_users_df = pd.merge(df_total_users, df_success_users, on='USER_ID', how='left').reset_index()
```

```python
print(joined_users_df)
```

```python
# Создадим столбец SUCCESS_RATE, где будет расчет процента успешных задач для каждого USER_ID
joined_users_df['SUCCESS_RATE'] = (
    joined_users_df['SUCCESS_TASK_ID'] / joined_users_df['TOTAL_TASK_ID'] * 100
).round(2)
```

```python
joined_users_df = joined_users_df.sort_values(by='SUCCESS_RATE', ascending=False)
```

```python
# Добавим Фамилию пользователя для нагладности
last_name_df = df_users[['USER_ID', 'LAST_NAME']].copy()

joined_users_df = pd.merge(
    joined_users_df, 
    last_name_df,
    on='USER_ID', 
    how='inner'
).reset_index(drop=True)  # drop=True чтобы не добавлять старый индекс как столбец
```

## Задания № 2:
### Собрать фамилии менеджеров у кого показатели отработки задач (% неуспешных задач) по одной из категорий бизнеса хуже, чем в среднем у всех менеджеров по всем категориям.

```python
print(df.head())
```

```python
# Подсчитываем количество неуспешных задач по каждой категории для каждого пользователя
unsuccess_df = df.query("STATUS == 'fail'").groupby(['USER_ID', 'CATEGORY']).agg({'TASK_ID': 'count'}).rename(columns={'TASK_ID': 'FAILED_TASKS'})
```

```python
print(unsuccess_df)
```

```python
# Подсчитываем общее количество задач
total_tasks = df.groupby(['USER_ID', 'CATEGORY']).agg({'TASK_ID': 'count'}).rename(columns={'TASK_ID': 'TOTAL_TASKS'})
```

```python
print(total_tasks)
```

```python
# самопроверка
print(df.query("USER_ID == 1 and STATUS == 'fail'").groupby(['USER_ID', 'CATEGORY']).agg({'TASK_ID': 'count'}))
print('-----------------------------------------------')
print(df.query("USER_ID == 1").groupby(['USER_ID', 'CATEGORY']).agg({'TASK_ID': 'count'}))
```

```python
# Объединяем результаты
result = unsuccess_df.join(total_tasks)

# Рассчитываем процент неуспешных задач
result['FAILURE_RATE'] = result['FAILED_TASKS'] / result['TOTAL_TASKS'] * 100

# Сбрасываем индекс, если нужно получить привычный DataFrame
result = result.reset_index()

# Выводим результат
print(result)
```

```python
# Группировка по категории и вычисление среднего
avg_per_category = result.groupby('CATEGORY')['FAILURE_RATE'].mean().reset_index()
avg_per_category.columns = ['CATEGORY', 'MEAN_FAILURE_RATE']
```

```python
display(avg_per_category)
```

```python
# Объединяем средние с исходными данными для сравнения
result = pd.merge(result, avg_per_category, on='CATEGORY')
result['LESS_MEAN'] = result['FAILURE_RATE'] > result['MEAN_FAILURE_RATE']

# Получаем фамилии менеджеров, чьи показатели хуже среднего
df_less_mean = result[result['LESS_MEAN']][['USER_ID', 'CATEGORY', 'FAILURE_RATE', 'MEAN_FAILURE_RATE', 'LESS_MEAN']]
```

```python
display(df_less_mean)
```

```python
# Добавим Фамилию пользователя для нагладности
df_less_mean = pd.merge(
    df_less_mean, 
    last_name_df,
    on='USER_ID', 
    how='inner'
).reset_index(drop=True)  # drop=True чтобы не добавлять старый индекс как столбец
```

```python
df_less_mean = df_less_mean[['USER_ID', 'LAST_NAME']].drop_duplicates().reset_index(drop=True)
```

### Используя библиотеку BeautifulSoup можно осуществить парминг данных прямо с сайта www.tks.ru

```python
# Карточка контрагента лежит в <div class="database-card__content">
```

```python
import requests
from bs4 import BeautifulSoup

# URL для парсинга
url = 'https://www.tks.ru/db/carrier/1183/'

# Получаем HTML-код страницы
response = requests.get(url)

# Проверяем, что запрос успешен
if response.status_code == 200:
    # Парсим HTML-код
    soup = BeautifulSoup(response.text, 'html.parser')

    # Пример: найти заголовок страницы
    title = soup.find('title').text
    print(f'Заголовок страницы: {title}')

    # найти все элементы страницы
    data = soup.find_all('div', class_='database-card__content')
    for item in data:
        print(item.text.strip())  # Выводим карточку партнера
else:
    print(f'Ошибка при запросе страницы: {response.status_code}')
```

### Используя библиотеку BeautifulSoup можно осуществить парминг данных прямо с сайта www.rbc.ru

```python
import requests
from bs4 import BeautifulSoup

# URL для парсинга
url = 'https://companies.rbc.ru/search/?category_id=542'

# Отправляем GET-запрос к странице
response = requests.get(url)

# Проверяем успешность запроса
if response.status_code == 200:
    # Парсим контент страницы с помощью BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Ищем все элементы, содержащие информацию о компаниях
    # Необходимо уточнить, какие именно теги и атрибуты содержат нужную информацию
    company_element = soup.find('a', class_='company-name-highlight')
    if company_element:
        company_name = company_element.get_text()
    else:
        company_name = None
    category = soup.find('div', class_='category-breadcrumb').find('span', class_='category-breadcrumb__item').text
    link = soup.find('a', class_='company-name-highlight')['href']
    # Здесь измените селектор, если необходимо
    # Вывод результатов
    print(f"Наименование компании: {company_name}")
    print(f"Категория: {category}")
    print(f"Ссылка: {link}")

else:
    print(f'Ошибка при запросе: {response.status_code}')
```
